# Load Tbox

```python
g = Graph()

# Load standards-compliant ontology
g.parse(os.path.join(base_dir,'FederatedTeam/tbox_standards.ttl'), format='ttl')

# Bind standard namespaces
g.bind('tb', 'http://www.semanticweb.org/acraf/ontologies/2024/healthmesh/tbox#')
g.bind('ab', 'http://www.semanticweb.org/acraf/ontologies/2024/healthmesh/abox#')
g.bind('dcat', 'http://www.w3.org/ns/dcat#')
g.bind('csvw', 'http://www.w3.org/ns/csvw#')
g.bind('dcterms', 'http://purl.org/dc/terms/')
g.bind('schema', 'http://schema.org/')
g.bind('odrl', 'http://www.w3.org/ns/odrl/2/')
g.bind('dqv', 'http://www.w3.org/ns/dqv#')
g.bind('prov', 'http://www.w3.org/ns/prov#')
```

# Populate ABOX

Util function for populating ABOX

```python
def add_jsonld_instances(graph, path):
    # Adds JSON-LD instances to the graph
    with open(path, 'r') as f:
        json_ld_data = json.loads(f.read())
        instances = Graph().parse(data=json_ld_data, format='json-ld')
        graph += instances
    
    return graph
```
