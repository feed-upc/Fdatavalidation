```python
import pandas as pd
from datetime import datetime, timedelta
import numpy as np
import os

# Load the UPENN dataset
file_path = "/home/acraf/psr/Fdatavalidation-1/DataProductLayer/DataProduct/Data/Explotation/UPENN-GBM_clinical_info_v2.1.csv"

# Read the CSV
df = pd.read_csv(file_path)

# Generate synthetic timestamps for each row
# Starting from now and going back in time with random intervals
base_time = datetime.now()
num_rows = len(df)

# Create random time offsets (between 0 and 60 days back)
random_days_back = np.random.uniform(0, 60, num_rows)
timestamps = [base_time - timedelta(days=float(days)) for days in random_days_back]

# Add timestamp column
df['ts'] = timestamps

# Save with new filename including "timestamp"
output_path = "/home/acraf/psr/Fdatavalidation-1/DataProductLayer/DataProduct/Data/Explotation/UPENN-GBM_clinical_info_v2.1_timestamp.csv"
df.to_csv(output_path, index=False)

print(f"Saved dataset with timestamps to {os.path.basename(output_path)}")
print(f"Dataset shape: {df.shape}")
print(f"Timestamp range: {df['ts'].min()} to {df['ts'].max()}")
print(f"Sample timestamps:")
print(df[['ID', 'ts']].head(10))
```
